```python
import pandas as pd

df = pd.read_csv("club_signups.csv")
df.head()
```

```python
print(df.shape)
print(df.dtypes)

for col in ["faculty", "club", "city", "fee_paid"]:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False))
```

```python
for col in ["faculty", "club", "city"]:
    df[col] = df[col].str.strip().str.lower()

for col in ["faculty", "club", "city"]:
    print(df[col].value_counts(), "\n")
```

```python
faculty_map = {
    "met": "MET",
    "media engineering and technology": "MET",
    "iet": "IET",
    "information engineering and technology": "IET",
    "ems": "EMS",
    "engineering and materials science": "EMS",
    "pharmacy": "Pharmacy",
    "pharma": "Pharmacy",
    "management": "Management",
}

club_map = {
    "robotics": "Robotics",
    "debate": "Debate",
    "debating": "Debate",
    "debate club": "Debate",
    "music": "Music",
    "football": "Football",
    "soccer": "Football",
    "chess": "Chess",
    "chess club": "Chess",
}

city_map = {
    "cairo": "Cairo",
    "giza": "Giza",
    "el giza": "Giza",
    "alexandria": "Alexandria",
    "alex": "Alexandria",
}

df["faculty"] = df["faculty"].map(faculty_map)
df["club"] = df["club"].map(club_map)
df["city"] = df["city"].map(city_map)
```

```python
assert set(df["faculty"]) <= {"MET", "IET", "EMS", "Pharmacy", "Management"}
assert set(df["club"]) <= {"Robotics", "Debate", "Music", "Football", "Chess"}
assert set(df["city"]) <= {"Cairo", "Giza", "Alexandria"}
```

```python
for col in ["faculty", "club", "city"]:
    print(df[col].value_counts(dropna=False), "\n")
```

```python
df["fee_paid"].value_counts()
```

```python
fee_map = {
    "y": True, "yes": True, "true": True, "1": True,
    "n": False, "no": False, "false": False, "0": False,
}

df["fee_paid"] = df["fee_paid"].astype(str).str.strip().str.lower().map(fee_map)

assert df["fee_paid"].isna().sum() == 0
df["fee_paid"] = df["fee_paid"].astype(bool)
assert df["fee_paid"].dtype == bool

df["fee_paid"].value_counts()
```

```python
df["signed_up_at"].head(10)
```

```python
s = df["signed_up_at"].str.strip()

iso = pd.to_datetime(s, format="%Y-%m-%d %H:%M", errors="coerce")
dmy = pd.to_datetime(s, format="%d/%m/%Y %H:%M", errors="coerce")

print("ISO rows:", iso.notna().sum())
print("day/month rows:", dmy.notna().sum())

df["signed_up_at"] = iso.fillna(dmy)

assert df["signed_up_at"].isna().sum() == 0
print(df["signed_up_at"].dtype)
print(df["signed_up_at"].min(), "to", df["signed_up_at"].max())
```

```python
df.to_csv("club_signups_clean.csv", index=False)
```
